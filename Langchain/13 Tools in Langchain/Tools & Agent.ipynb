```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import AgentType
from langchain.agents import Tool
from langchain.agents import initialize_agent
```

```python
# Load environment variables (for GOOGLE_API_KEY)
load_dotenv()
```

```python
# Initialize Gemini Pro model
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-pro",
    google_api_key=os.getenv("GOOGLE_API_KEY"),
    temperature=0.7
)
```

```python
# --- Tool Functions ---
def add_numbers_tool(input_str: str) -> str:
    """Adds two numbers together"""
    num1, num2 = map(float, input_str.split(','))
    return str(num1 + num2)
```

```python
def reverse_string_tool(text: str) -> str:
    """Reverses a given string."""
    return text[::-1]
```

```python
# --- Create LangChain Tool Objects ---
add_tool = Tool(
    name="AddNumbers",
    func=add_numbers_tool,
    description="Useful for adding two numbers."
)
```

```python
reverse_tool = Tool(
    name="ReverseString",
    func=reverse_string_tool,
    description="Useful for reversing any given string. Input should be a single string."
)
```

```python
# --- List of tools available to the agent ---
tools = [add_tool, reverse_tool]
```

```python
# --- Create Agent ---
agent = initialize_agent(
    tools=tools,
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)
```

```python
# --- Run Agent Queries ---
if __name__ == "__main__":
    print("\n--- Query 1 ---")
    print(agent.run("Add 30 and 45"))

    print("\n--- Query 2 ---")
    print(agent.run("Reverse the string: LangChain is powerful"))
```
